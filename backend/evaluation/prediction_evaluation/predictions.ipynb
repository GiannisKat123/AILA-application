### Pipeline

```python
from langgraph.graph import StateGraph
from langgraph.checkpoint.memory import MemorySaver
from typing_extensions import TypedDict, List, Annotated, Dict
import operator, uuid, ast
from concurrent.futures import ThreadPoolExecutor, as_completed
from openai.cli._errors import OpenAIError
from backend.database.core.funcs import get_prompt
from langchain_openai import ChatOpenAI
from backend.database.config.config import settings
from llama_index.core.base.base_retriever import BaseRetriever
from langchain_core.documents.base import Document


model = ChatOpenAI(model=settings.OPEN_AI_MODEL,api_key=settings.OPENAI_API_KEY,temperature=0.7)
top_k=10

class AgentState(TypedDict):
    user_query:str
    summarized_context:str
    saerch_results:str
    questions:List[str]
    query_classification:Annotated[Dict[str,List[str]],operator.or_]
    retrieved_docs:Annotated[Dict[str,List],operator.or_]


def retrieving_docs(query:str,indexes:List[BaseRetriever]):
    retrieved_nodes = []
    for index in indexes:
        index = retriever_map[index]
        nodes = index.retrieve(query)
        retrieved_nodes.append([Document(page_content=node.text,metadata=node.metadata) for node in nodes])
    
    documents = [node for index_nodes in retrieved_nodes for node in index_nodes]
    pairs = [(query,doc.page_content) for doc in documents]
    scores = reranker_model.predict(pairs)
    scored_docs = list(zip(scores,documents))
    scored_docs.sort(reverse=True,key=lambda x:x[0])
    reranked_docs = scored_docs[:top_k]
    return [[node.page_content,node.metadata,float(score)] for score,node in reranked_docs]

def query_classification(state,level:int):
    prompt = get_prompt(prompt_name='query_classification_on_index')
    response = model.invoke(prompt.format(query=state['questions'][level]))
    response_content = str(response.content).strip()

    res = response_content.split("Output:")
    if len(res) > 1:
        res = res[1]
    else:
        res = res[0]

    if isinstance(res, list) and isinstance(res[0], str):
        res = res[0]
    
    try:
        categories = ast.literal_eval(res)

        if len(categories) > 0:
            indexes = []
            for category in categories:
                if category == 'GDPR':
                    indexes += ["gdpr_index_recall_retriever","gdpr_index_precision_retriever"]
                if category == 'Greek Penal Code':
                    indexes += ["gpc_index_recall_retriever","gpc_index_precision_retriever"]
                if category == 'Specific Legal Cases':
                    indexes += ["law_cases_index_recall_retriever","law_cases_index_precision_retriever"]
                if category == 'Phishing Scenarios':
                    indexes += ["phishing_retriever"]
            state['query_classification'] = {level:[state['questions'][level],indexes]}
        else: state['query_classification'] = {level:[state['questions'][level],None]}

    except Exception as e:
        state['query_classification'] = {level:[state['questions'][level],None]}

    return {'query_classification':state['query_classification']}


def run_classifications_parallel(state):
    levels = [0,1,2]
    results = {}

    def classify(level):
        try:
            result = query_classification(state, level)
            return level, result
        except Exception as e:
            return level, {}
    
    with ThreadPoolExecutor(max_workers=3) as executor:
        futures = {executor.submit(classify, level): level for level in levels}
        for future in as_completed(futures):
            level, result = future.result()
            results[level] = result
    
    combined = {}
    for i in range(3):
        combined[i] = results[i]['query_classification'][i]
    state['query_classification'] = combined
    return {'query_classification': state['query_classification']}


def query_rewriting(state):
    prompt = get_prompt(prompt_name='query_rewritter')

    retries = 3
    for _ in range(retries):
        try:
            response = model.invoke(prompt.format(query = state['user_query']))

            response_content = str(response.content).strip()
            res = response_content.split("Output:")
            res = ast.literal_eval(res[1])
            questions = {0:state['user_query'],1:res[0],2:res[1]}

            state['questions'] = questions
            return {'questions':questions}
        
        except OpenAIError:
            raise RuntimeError("Exceeded current quota, please contact the administrator.")  # ✅ Fixed
        
        except Exception as e:
            continue  
    
    raise RuntimeError("❌ Failed to rewrite query after multiple attempts.")


def run_retrievals_parallel(state):
    levels = [0,1,2]
    results = {}

    def retrieve(level):
        return level, retrieve_docs(state, level)

    with ThreadPoolExecutor(max_workers=3) as executor:
        futures = {executor.submit(retrieve, level): level for level in levels}
        for future in as_completed(futures):
            level, result = future.result()
            results[level] = result

    state['retrieved_docs'] = results
    return {'retrieved_docs': state['retrieved_docs']}


def retrieve_docs(state,level):
    retrieved_documents = retrieving_docs(state['questions'][0],state['query_classification'][level][1]) if state['query_classification'][level][1] else None
    state['retrieved_docs'][level] = retrieved_documents
    return {level:state['retrieved_docs'][level]}

def initialize_workflow():
    """
    Build and compile the LangGraph workflow:

        query_rewriting → parallel_classification → parallel_retrieval → get_context

    Returns:
        Any: Compiled app instance with an in-memory checkpointer.
    """
    workflow = StateGraph(AgentState)

    ## Query re-writing
    workflow.add_node('query_rewriting',query_rewriting)
    ## Query Categorization of query and variants
    workflow.add_node('parallel_classification',run_classifications_parallel)
    ## Document Retrieval
    workflow.add_node('parallel_retrieval',run_retrievals_parallel)
    ## Document Aggregation and Response

    ## Query re-writing -> Query Categorization
    workflow.add_edge("query_rewriting","parallel_classification")
    # ## Query Categorization -> Retrieval Documents
    workflow.add_edge("parallel_classification","parallel_retrieval")
    # ## Retrieval Documents -> Document Aggregation and Response

    workflow.set_entry_point("query_rewriting")
    checkpointer = MemorySaver()
    app = workflow.compile(checkpointer = checkpointer)

    return app

app = initialize_workflow()

def rag_pipeline(query:str,app):
    config = {"configurable": {"thread_id": f"{uuid.uuid4()}"}}
    result = app.invoke({
        "user_query":query,
        "questions": [],  # <-- ADD THIS
        "query_classification": {},  # <-- FIXED
        "retrieved_docs": {},  # <-- ADD THIS
        "context": {},  # <-- ALREADY GOOD
    }, config)


    return {"query":query,
        'result':result,
        }

def run_full_pipeline(query:str,app):   
    rag_search = rag_pipeline(query,app)
    return rag_search['result']['retrieved_docs'][0][0] + rag_search['result']['retrieved_docs'][1][1] + rag_search['result']['retrieved_docs'][2][2]
```

```python
folders = os.listdir(new_path + '//evaluation//files')
input_paths = {f:[new_path+f'//evaluation//files//{f}'] for f in folders}
```

```python
queries_files = [[f"{new_path}/evaluation/synthetic_data_new/synthetic_data_{folder}.csv",
                f"{new_path}/evaluation/synthetic_data_new/more_synthetic_data_{folder}.csv"] for folder in folders]

questions = []
corpus_texts = []
corpus_texts_id = []
grouping = []

for file,theme in zip(queries_files,folders):
    for csv_file in file:
        
        df = pd.read_csv(csv_file)
        questions += df['question'].to_list()
        corpus_text_list = df['corpus_id'].to_list()
        corpus_texts_id += corpus_text_list
        for corpus in corpus_text_list:
            grouping.append(theme)
            corpus = corpus.split('/')
            corpus_path = new_path + f'/evaluation/files/{theme}/{corpus[-1]}'
            with open(corpus_path,'r',encoding='utf-8') as f:
                text = f.read()
            corpus_texts.append(text)

corpus_id = np.unique(corpus_texts_id) 
corpus_id_dict = {corpus_id[i]:i for i in range(len(corpus_id))}

res = []

for i in range(len(questions)):
    res.append([questions[i],corpus_texts[i],corpus_id_dict[corpus_texts_id[i]],i,corpus_texts_id[i].split('.txt')[0].split('/')[-1],grouping[i]])

df = pd.DataFrame(res,columns=['anchor','positive','global_chunk_id','id','corpus_title','group'])
```
