```python
df = pd.read_csv('Benchmarks of Langchain\predictions_index_trained_embeddings_new_with_cohere.csv')
predictions = df['predictions'].to_list()

new_predictions = [ast.literal_eval(pred) for pred in predictions]

all_scores = []

for pred in new_predictions:
    if len(pred) == 2:
        continue
    else:
        for p in pred:
            all_scores.append(p[2])

import matplotlib.pyplot as plt
import seaborn as sns

sns.histplot(all_scores, bins=50, kde=True)
plt.title("Reranker Score Distribution")
plt.xlabel("Score")
plt.ylabel("Frequency")
plt.show()

import numpy as np

mean = np.mean(all_scores)
std = np.std(all_scores)
percentile_25 = np.percentile(all_scores, 25)
percentile_50 = np.percentile(all_scores, 50)
percentile_75 = np.percentile(all_scores, 75)

print(f"Mean: {mean:.3f}, Std: {std:.3f}")
print(f"25th: {percentile_25:.3f}, 50th: {percentile_50:.3f}, 75th: {percentile_75:.3f}")
```
